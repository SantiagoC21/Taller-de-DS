```python
import os
import pysd
import matplotlib.pyplot as plt
```

```python
def funcAccidentesTransito():
    modelo = pysd.read_vensim('FORRESTER-Satisfaccion_autoridades.mdl')
    valores = modelo.run(return_columns = ['Accidentes de tránsito'])
    tabla = valores.head(6)
    print(tabla)
    valores.plot()
    plt.ylabel('Cantidad')
    plt.xlabel('Años')
    plt.legend(loc='upper center')
    plt.legend(loc='center left')
    plt.title("Accidentes de tránsito")
    plt.show()
    return plt
```

```python
funcAccidentesTransito()
```
