```python
import pysd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
model = pysd.read_vensim('frecuencia de mantenimiento.mdl')
results = model.run()
```

```python
disponibilidad_de_la_flota = results["Disponibilidad de la flota"]
objetivo_disponibilidad_de_la_flota = 30
discrepancia_disponibilidad_de_la_flota = 30 - disponibilidad_de_la_flota

# Crear la figura
plt.figure(figsize=(10, 6))

# Graficar la Disponibilidad de la Flota
plt.plot(disponibilidad_de_la_flota, label='Disponibilidad de la Flota', color='blue')

# Graficar la Discrepancia de la Flota
plt.plot(discrepancia_disponibilidad_de_la_flota, label='Discrepancia de la Flota', color='green')

# Graficar el objetivo (constante)
plt.axhline(y=objetivo_disponibilidad_de_la_flota, color='red', linestyle='--', label='Objetivo (30)')

# Añadir título y etiquetas
plt.title('Disponibilidad y Discrepancia de la Flota')
plt.xlabel('Semanas')
plt.ylabel('Valor')

# Activar la cuadrícula
plt.grid(True)

# Mostrar la leyenda
plt.legend()

# Mostrar el gráfico
plt.show()
```

```python
cantidad_de_vehiculos_en_operacion = results["Cantidad de vehiculos en operacion"]
objetivo_cantidad_de_vehiculos_en_operacion = 150
discrepancia_cantidad_de_vehiculos_en_operacion = 150 - cantidad_de_vehiculos_en_operacion

# Crear la figura
plt.figure(figsize=(10, 6))

# Graficar la Disponibilidad de la Flota
plt.plot(cantidad_de_vehiculos_en_operacion, label='Cantidad de vehiculos en operacion', color='blue')

# Graficar la Discrepancia de la Flota
plt.plot(discrepancia_cantidad_de_vehiculos_en_operacion, label='Discrepancia de la Cantidad de vehiculos en operacion', color='green')

# Graficar el objetivo (constante)
plt.axhline(y=objetivo_cantidad_de_vehiculos_en_operacion, color='red', linestyle='--', label='Objetivo (30)')

# Añadir título y etiquetas
plt.title('Cantidad de vehiculos en operacion')
plt.xlabel('Semanas')
plt.ylabel('Valor')

# Activar la cuadrícula
plt.grid(True)

# Mostrar la leyenda
plt.legend()

# Mostrar el gráfico
plt.show()
```
