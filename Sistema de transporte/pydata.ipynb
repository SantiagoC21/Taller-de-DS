```python
df = pd.read_csv("data_modificada_final.csv", delimiter=',')
df
```

```python
variables_y = [
    "Disponibilidad de la flota",
    "Vehiculos que regresan de mantenimiento",
    "Vehiculos que se retiran por mantenimiento",
    "Vehiculos asignados a operar"
]

plt.figure()

for variable in variables_y:
    plt.plot(df["Weeks"], df[variable], label=variable)

plt.xlabel("Semana")
plt.ylabel("Valor")
plt.title("Nivel: Disponibilidad de la flota")
plt.legend(loc="best", bbox_to_anchor=(1, 0.5))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
variables_y = [
    "cantidadDeVehiculosEnOperacion",
    "vehiculosAsignadosAOperar",
    "vehiculosDevueltosAlDeposito"
]

plt.figure()

for variable in variables_y:
    plt.plot(df["sem"], df[variable], label=variable)

plt.xlabel("Semana")
plt.ylabel("Valor")
plt.title("Nivel: Disponibilidad de la flota")
plt.legend(loc="best", bbox_to_anchor=(1, 0.5))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots()

# Primer eje Y (izquierda)
ax1.plot(df["sem"], df["disponibilidadDeTalleres"], label="disponibilidadDeTalleres", color="blue")
ax1.set_xlabel("Semana")
ax1.set_ylabel("Disponibilidad de Talleres", color="blue")
ax1.tick_params(axis='y', labelcolor="blue")

# Segundo eje Y (derecha)
ax2 = ax1.twinx()
ax2.plot(df["sem"], df["cantidadDePlazasSin"], label="cantidadDePlazasSin", color="orange")
ax2.plot(df["sem"], df["cantidadDePlazasApartadasPorMantenimiento"], label="cantidadDePlazasApartadasPorMantenimiento", color="green")
ax2.set_ylabel("Cantidad de Plazas", color="green")
ax2.tick_params(axis='y', labelcolor="green")

# Título y rejilla
plt.title("Nivel: Disponibilidad de la flota")
fig.tight_layout()
plt.grid(True)
plt.show()
```

```python
df_modificado = df.copy()
df_modificado["Disponibilidad de la flota"] *= (1 - 0.005 * df_modificado["Weeks"])
df_modificado["Vehiculos que regresan de mantenimiento"] *= (1 - 0.01 * df_modificado["Weeks"])
df_modificado["Vehiculos que se retiran por mantenimiento"] *= (1 + 0.01 * df_modificado["Weeks"])
df_modificado["Vehiculos asignados a operar"] *= (1 + 0.008 * df_modificado["Weeks"])

# Definir las variables a graficar y sus colores
variables_y = [
    "Disponibilidad de la flota",
    "Vehiculos que regresan de mantenimiento",
    "Vehiculos que se retiran por mantenimiento",
    "Vehiculos asignados a operar"
]
colores = ['blue', 'orange', 'green', 'red']

# Crear los subgráficos
fig, axs = plt.subplots(nrows=4, sharex=True, figsize=(10, 8))

for i, var in enumerate(variables_y):
    axs[i].plot(df_modificado["Weeks"], df_modificado[var], color=colores[i])
    axs[i].set_ylabel("")
    axs[i].grid(True)

plt.xlabel("Semana")
plt.suptitle("Nivel: Disponibilidad de la flota", fontsize=14)
fig.legend(variables_y, loc="upper left", bbox_to_anchor=(1.0, 0.5))
plt.tight_layout(rect=[0, 0.03, 0.85, 0.95])
plt.show()
```

```python
df_mod = df.copy()

# Aplicar modificaciones progresivas por semana
df_mod["Cantidad de vehiculos en operacion"] *= (1 + 0.01 * df_mod["Weeks"])
df_mod["Vehiculos asignados a operar"] *= (1 + 0.008 * df_mod["Weeks"])
df_mod["Vehiculos devueltos al deposito"] *= (1 - 0.005 * df_mod["Weeks"])

# Definir variables y colores
variables_y = [
    "Cantidad de vehiculos en operacion",
    "Vehiculos asignados a operar",
    "Vehiculos devueltos al deposito"
]
colores = ['blue', 'orange', 'green']

# Crear los subgráficos
fig, axs = plt.subplots(nrows=3, sharex=True, figsize=(10, 8))

for i, var in enumerate(variables_y):
    axs[i].plot(df_mod["Weeks"], df_mod[var], color=colores[i])
    axs[i].set_ylabel("") 
    axs[i].grid(True)

plt.xlabel("Semana")
plt.suptitle("Nivel: Cantidad de vehículos en operación", fontsize=14)
fig.legend(variables_y, loc="upper left", bbox_to_anchor=(1.0, 0.5))
plt.tight_layout(rect=[0, 0.03, 0.85, 0.95])
plt.show()
```

```python
df_talleres_mod = df.copy()

# Aplicar mejoras progresivas a lo largo de las semanas
df_talleres_mod["Disponibilidad de talleres"] *= (1 + 0.015 * df_talleres_mod["Weeks"])
df_talleres_mod["Cantidad de plazas sin asignacion de mantenimiento"] *= (1 - 0.015 * df_talleres_mod["Weeks"])
df_talleres_mod["Cantidad de plazas apartadas por mantenimiento"] *= (1 - 0.02 * df_talleres_mod["Weeks"])

# Variables a graficar y sus colores
variables_y = [
    "Disponibilidad de talleres",
    "Cantidad de plazas sin asignacion de mantenimiento",
    "Cantidad de plazas apartadas por mantenimiento"
]
colores = ['blue', 'orange', 'green']

# Crear subplots para graficar los datos modificados
fig, axs = plt.subplots(nrows=3, sharex=True, figsize=(10, 8))

for i, var in enumerate(variables_y):
    axs[i].plot(df_talleres_mod["Weeks"], df_talleres_mod[var], color=colores[i])
    axs[i].set_ylabel("") 
    axs[i].grid(True)

plt.xlabel("Semana")
plt.suptitle("Nivel: Disponibilidad de talleres", fontsize=14)
fig.legend(variables_y, loc="upper left", bbox_to_anchor=(1.0, 0.5))
plt.tight_layout(rect=[0, 0.03, 0.85, 0.95])
plt.show()
```
